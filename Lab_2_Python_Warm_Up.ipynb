### 1. Print something

`print()` shows whatever is inside the parentheses. Text goes in quotes.

```python
print("Hello, Georgetown")
```

### 2. Variables: labeled boxes

A **variable** stores a value under a name. `=` means "store this," not "equals."

```python
price = 120
units = 3

price * units
```

The last line of a cell is shown automatically, so you do not always need `print()`.

### 3. Math: a calculator that remembers

Here: revenue from an order after a 15% discount.

```python
discount = 0.15
revenue = price * units * (1 - discount)

print("Revenue:", revenue)
```

### 4. Text needs quotes

Text in quotes is a **string**. Without quotes, Python thinks it is a variable name.

```python
channel = "Referral"

print("Customer came from", channel)
```

### 5. Lists: several values together

A **list** stores several values under one name, inside square brackets. Here, the number of orders from five customers. `max()` finds the largest value, `sum()` adds them up, and `len()` counts them.

```python
orders = [2, 8, 1, 5, 3]

print("Largest:", max(orders))
print("Average:", sum(orders) / len(orders))
```

### 6. Dictionaries: values with labels

A **dictionary** stores values under labels, called **keys**, inside curly braces. Each pair is written `key: value`. You get a value back by its key, in square brackets.

```python
customer = {"name": "Ana", "channel": "Referral", "orders": 8}

customer["channel"]
```

A list can hold dictionaries. This is how a conversation with an AI model is sent from code: a list of messages, each with a role and its content.

```python
messages = [
    {"role": "user", "content": "What is a large language model?"},
    {"role": "assistant", "content": "A model that predicts the next word."},
]

messages[1]["content"]
```

### 7. If: make a decision

`if` runs the indented line only when the condition is met. `else` covers everything else.

```python
orders_this_year = 8

if orders_this_year > 5:
    print("Frequent buyer")
else:
    print("Occasional buyer")
```

### 8. pandas: a spreadsheet inside Python

**pandas** is a library for working with data tables: rows, columns, totals, and filters, like Excel but with code. `import` loads it, and `pd` is its nickname. The `{...}` inside `pd.DataFrame` is a dictionary: each key is a column name, each value is that column's list.

```python
# load the pandas library, nicknamed pd
import pandas as pd

# build a table with two columns: channel and customers
table = pd.DataFrame({"channel": ["Email", "Referral"], "customers": [180, 150]})

# show the table
table
```
